```python
# AO* Search Algorithm

graph = {
    'A': [[('B', 1), ('C', 1)], [('D', 1)]],
    'B': [[('E', 1)]],
    'C': [[('F', 1)]],
    'D': [[('G', 1)]],
    'E': [],
    'F': [],
    'G': []
}

heuristic = {
    'A': 3,
    'B': 2,
    'C': 2,
    'D': 1,
    'E': 0,
    'F': 0,
    'G': 0
}

solution = {}

def ao_star(node):
    # Goal node
    if not graph[node]:
        return heuristic[node]

    best_cost = float('inf')
    best_option = None

    # Check every AND/OR option
    for option in graph[node]:

        cost = 0

        # AND: all nodes in this option must be solved
        for child, edge_cost in option:
            cost += edge_cost + ao_star(child)

        if cost < best_cost:
            best_cost = cost
            best_option = option

    solution[node] = best_option
    heuristic[node] = best_cost

    return best_cost


# Start AO* Search
cost = ao_star('A')

print("Optimal Cost:", cost)
print("Solution Graph:")

for node, option in solution.items():
    print(node, "->", [child for child, cost in option])
```
