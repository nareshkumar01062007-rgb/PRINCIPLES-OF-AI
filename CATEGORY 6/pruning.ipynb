```python
def alpha_beta(depth, node, maximizing, values, alpha, beta):

    if depth == 2:
        return values[node]

    if maximizing:
        best = float('-inf')

        for i in range(2):
            score = alpha_beta(depth + 1, node * 2 + i,
                               False, values, alpha, beta)

            best = max(best, score)
            alpha = max(alpha, best)

            if alpha >= beta:
                break

        return best

    else:
        best = float('inf')

        for i in range(2):
            score = alpha_beta(depth + 1, node * 2 + i,
                               True, values, alpha, beta)

            best = min(best, score)
            beta = min(beta, best)

            if alpha >= beta:
                break

        return best


values = [3, 5, 2, 9]

result = alpha_beta(0, 0, True, values,
                    float('-inf'), float('inf'))

print("Best score:", result)
```
